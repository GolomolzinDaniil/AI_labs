```python
import os, cv2
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
import cv2
cv2.setNumThreads(0)
cv2.utils.logging.setLogLevel(cv2.utils.logging.LOG_LEVEL_ERROR)
import numpy as np


TARGET = 'valid'
TARGET_LABELS = 'valid_labels'

def unpack(X_path: str, y_path: str):
    '''Вытаскивает пути к изображениям'''
    X, y = [], []
    for img1_path, lbl_path in zip(sorted(os.listdir(X_path)), sorted(os.listdir(y_path))):
        X.append(os.path.join(X_path, img1_path))
        y.append(os.path.join(y_path, lbl_path))
    return X, y
```

```python
path_data = r'/home/d.golomolzin/.cache/kagglehub/datasets/balraj98/massachusetts-roads-dataset/versions/1/tiff'

X_train_path = os.path.join(path_data, TARGET)
y_train_path = os.path.join(path_data, TARGET_LABELS)

X, y = unpack(X_train_path, y_train_path)
```

```python
size_crop = 500
num_show = 2
h, w = 1500, 1500

count_total = 0


for path1, path2 in zip(X, y):
    
    img1 = cv2.imread(path1)
    img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
    
    img2 = cv2.imread(path2, cv2.IMREAD_GRAYSCALE)

    count_total += 1
    
    for i in range(0, h, size_crop):
        for j in range(0, w, size_crop):

            image_crop = img1[i : i+size_crop, j : j+size_crop, :]
            label_crop = img2[i : i+size_crop, j : j+size_crop]

            # if np.mean(image_crop > 240) > 0.8 or np.mean(label_crop > 127) < 0.01: continue


            base_name = os.path.splitext(os.path.basename(path1))[0]

            save_image = f"/home/d.golomolzin/AI_labs/2lab/my_dataset/{TARGET}/{base_name}_{i}_{j}.tiff"
            cv2.imwrite(save_image, cv2.cvtColor(image_crop, cv2.COLOR_RGB2BGR))

            save_label = f'/home/d.golomolzin/AI_labs/2lab/my_dataset/{TARGET_LABELS}/{base_name}_{i}_{j}.tif'
            cv2.imwrite(save_label, label_crop)
            

    # if count_total == num_show: break
print(count_total)
```
